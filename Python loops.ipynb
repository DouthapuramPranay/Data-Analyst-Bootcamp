# FOR LOOP

```python
for i in range(5):
    print(i)
```

```python
num=[1,2,3,4,5]
for i in num:
    if i < 4:
        print(i)
    else:
        print("DONE")
        break
```

```python
num=[1,2,3,4,5]
for i in num:
    print(i+i)
```

```python
dict={'name':'pranay','age':21,'height':6}
```

```python
for i in dict.values():
    print(i)
```

```python
for key,value in dict.items():
    print(key,'->',value)
```

```python
i=['Pranay','Arvind','Narendra']
j=['ice cream','choclate','fruits']

for k in i:
    for l in j:
        print(k,"eats",l)
```

# WHILE LOOP

```python
num=0
while num<5:
    print(num)
    num+=1
```

```python
num=0
while num<5:
    print(num)
    if num==3:
        break
    num+=1
```

```python
num=0
while num<5:
    print(num)
    if num==3:
        break
    num+=1
else:
    print("No longer responding")
```

```python
num=0
while num<5:
    print(num)
    if num==6:
        break
    num+=1
else:
    print("No longer responding")
```

```python
num=0
while num<5:
    print(num)
    if num==3:
        continue
    num+=1
else:
    print("No longer responding")
```

```python
num=0
while num<5:
    num+=1
    if num==3:
        continue
    print(num)
else:
    print("No longer responding")
```
